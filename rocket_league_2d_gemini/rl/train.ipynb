```python
from stable_baselines3 import PPO
from rl.env import RocketSoccerEnv

class RocketSoccerTrainer:
    def __init__(self, model_version="v1", timesteps=300000):
        self.model_version = model_version
        self.timesteps = timesteps
        # Saves directly into your mounted Google Drive!
        self.save_dir = os.path.join("rl", "versions")
        self.env = None
        self.model = None

    def setup(self):
        print("1. Initializing Environment...")
        os.makedirs(self.save_dir, exist_ok=True)
        self.env = RocketSoccerEnv()

    def build_model(self):
        print("2. Building PPO Brain...")
        self.model = PPO(
            policy="MlpPolicy", 
            env=self.env, 
            verbose=1, 
            learning_rate=0.0003,
            device="auto" # Colab will automatically use the NVIDIA GPU here!
        )

    def train(self):
        print(f"3. Starting Training for {self.timesteps} timesteps...")
        if self.model is None:
            self.build_model()
        self.model.learn(total_timesteps=self.timesteps)

    def save_model(self):
        save_path = os.path.join(self.save_dir, self.model_version)
        self.model.save(save_path)
        print(f"4. Success! Model securely saved to: {save_path}.zip")

# --- RUN THE TRAINING ---
# Start with 20000 to test, then change to 300000 or 500000 for the real brain!
trainer = RocketSoccerTrainer(model_version="v1", timesteps=20000)
trainer.setup()
trainer.build_model()
trainer.train()
trainer.save_model()
```
